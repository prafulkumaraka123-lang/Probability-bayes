## Question 1 — Random variables, PMF and PDF

A **random variable** is a numerical value assigned to each outcome of a random experiment.

- **Discrete random variable:** Takes a finite or countably infinite set of values.
  - Examples: number of heads in three coin tosses; number shown when rolling a die.
- **Continuous random variable:** Can take any value in an interval or range.
  - Examples: a person's height; the time taken to complete a task.

A **probability mass function (PMF)** gives the probability that a discrete random variable equals a particular value: \(p(x)=P(X=x)\).

A valid PMF must satisfy:
1. \(p(x)\geq 0\) for every possible value of \(x\).
2. \(\sum_x p(x)=1\), where the sum is over all possible values.

A **probability density function (PDF)** describes the probability distribution of a continuous random variable. Probabilities are areas under its curve:
\(P(a\leq X\leq b)=\int_a^b f(x)\,dx\).
A valid PDF is non-negative and its total integral over the full range is 1. For a continuous variable, the probability at one exact point is zero.

## Question 2 — Number of heads in three fair coin tosses

Let \(X\) be the number of heads. The eight equally likely outcomes and corresponding values are:

| Outcome | X |
|---|---:|
| HHH | 3 |
| HHT | 2 |
| HTH | 2 |
| HTT | 1 |
| THH | 2 |
| THT | 1 |
| TTH | 1 |
| TTT | 0 |

Therefore the PMF is:

| x | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| P(X = x) | 1/8 | 3/8 | 3/8 | 1/8 |

The probabilities add to \((1+3+3+1)/8=1\). Also,
\(P(X\geq 2)=P(X=2)+P(X=3)=3/8+1/8=1/2=0.5\).

```python
import matplotlib.pyplot as plt

x = [0, 1, 2, 3]
probabilities = [1/8, 3/8, 3/8, 1/8]

print("Sum of probabilities:", sum(probabilities))
print("P(X >= 2):", probabilities[2] + probabilities[3])

plt.bar(x, probabilities)
plt.xticks(x)
plt.xlabel("Number of heads (X)")
plt.ylabel("Probability")
plt.title("PMF: Number of heads in 3 fair coin tosses")
plt.ylim(0, 0.5)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## Question 3 — Expected value

The **expected value** is the long-run average value of a random variable over many repetitions. For a discrete random variable:

\[
E(X)=\sum_x xP(X=x)
\]

Given:

| x | P(X = x) | xP(X = x) |
|---:|---:|---:|
| 1 | 0.1 | 0.1 |
| 2 | 0.3 | 0.6 |
| 3 | 0.4 | 1.2 |
| 4 | 0.2 | 0.8 |

\[
E(X)=1(0.1)+2(0.3)+3(0.4)+4(0.2)
=0.1+0.6+1.2+0.8=\boxed{2.7}
\]

**Interpretation:** Over many repetitions, the average value of X approaches 2.7. It does not mean that X must equal 2.7 in an individual observation.

```python
values = [1, 2, 3, 4]
probs = [0.1, 0.3, 0.4, 0.2]

expected_value = sum(x * p for x, p in zip(values, probs))
print("E(X) =", expected_value)
```

## Question 4 — Properties of expectation and a game

**Linearity of expectation:**

\[
E(aX+b)=aE(X)+b
\]

Multiplying a random variable by a constant multiplies its expected value by that constant; adding a constant shifts its expected value by that constant.

\[
E(X+Y)=E(X)+E(Y)
\]

This property holds whether or not X and Y are independent.

The game costs ₹20. The gross winnings are ₹50 with probability 0.3 and ₹0 with probability 0.7.

Expected gross winnings:
\[
E(W)=50(0.3)+0(0.7)=₹15
\]

Expected profit to the player:
\[
E(\text{profit})=₹15-₹20=\boxed{-₹5}
\]

The player loses ₹5 on average per game. Hence, it is **not fair to the player**. A fair game would have expected net profit equal to zero.

```python
cost = 20
win_amount = 50
p_win = 0.3
expected_winnings = win_amount * p_win
expected_player_profit = expected_winnings - cost

print("Expected gross winnings: Rs.", expected_winnings)
print("Expected net profit for player: Rs.", expected_player_profit)
```

## Question 5 — Variance and standard deviation

**Variance** measures how spread out the values of a random variable are around its mean. **Standard deviation** is the positive square root of variance and is expressed in the same units as the variable.

\[
\operatorname{Var}(X)=E[(X-E(X))^2]
\]

Expanding the square gives the shortcut formula:

\[
\operatorname{Var}(X)=E(X^2)-[E(X)]^2
\]

Using Question 3:

\[
E(X^2)=1^2(0.1)+2^2(0.3)+3^2(0.4)+4^2(0.2)
\]
\[
=0.1+1.2+3.6+3.2=8.1
\]

Since \(E(X)=2.7\):

\[
\operatorname{Var}(X)=8.1-(2.7)^2=8.1-7.29=\boxed{0.81}
\]

\[
\sigma=\sqrt{0.81}=\boxed{0.9}
\]

```python
import numpy as np

values = np.array([1, 2, 3, 4], dtype=float)
probs = np.array([0.1, 0.3, 0.4, 0.2])

EX = np.sum(values * probs)
EX2 = np.sum((values ** 2) * probs)
variance = EX2 - EX**2
std_dev = np.sqrt(variance)

print("E(X):", EX)
print("E(X^2):", EX2)
print("Variance:", variance)
print("Standard deviation:", std_dev)
```

## Question 6 — Variance transformation and die simulation

For constants \(a\) and \(b\):

\[
\operatorname{Var}(aX+b)=a^2\operatorname{Var}(X)
\]

Adding a constant shifts all values but does not change their spread. Multiplying by \(a\) scales deviations by \(a\), so the variance is scaled by \(a^2\).

Given \(\operatorname{Var}(X)=5\):

\[
\operatorname{Var}(3X+7)=3^2(5)=\boxed{45}
\]
\[
\operatorname{Var}(-2X)=(-2)^2(5)=\boxed{20}
\]

For a fair six-sided die, the theoretical mean is \((1+2+3+4+5+6)/6=3.5\), and the theoretical variance is \(35/12\approx2.9167\).

```python
import numpy as np

rng = np.random.default_rng(42)
rolls = rng.integers(1, 7, size=100_000)

simulated_mean = np.mean(rolls)
# ddof=0 gives the population-style variance of the simulated outcomes
simulated_variance = np.var(rolls, ddof=0)

theoretical_mean = 3.5
theoretical_variance = 35 / 12

print(f"Simulated mean:       {simulated_mean:.5f}")
print(f"Theoretical mean:     {theoretical_mean:.5f}")
print(f"Simulated variance:   {simulated_variance:.5f}")
print(f"Theoretical variance: {theoretical_variance:.5f}")
```

## Question 7 — Covariance and sample covariance

**Covariance** measures how two variables vary together.

- **Positive covariance:** Larger values of one variable tend to occur with larger values of the other.
- **Negative covariance:** Larger values of one tend to occur with smaller values of the other.
- **Zero covariance:** There is no linear co-variation indicated by covariance; this does not necessarily mean the variables are independent.

Given \(x=[2,4,6,8]\) and \(y=[1,3,5,9]\):

\[
\bar{x}=5,\qquad \bar{y}=4.5
\]

| x | y | x − x̄ | y − ȳ | Product |
|---:|---:|---:|---:|---:|
| 2 | 1 | -3 | -3.5 | 10.5 |
| 4 | 3 | -1 | -1.5 | 1.5 |
| 6 | 5 | 1 | 0.5 | 0.5 |
| 8 | 9 | 3 | 4.5 | 13.5 |
| | | | **Sum** | **26** |

Sample covariance:

\[
s_{xy}=\frac{\sum (x_i-\bar{x})(y_i-\bar{y})}{n-1}
=\frac{26}{4-1}=\boxed{8.6667}
\]

The \(n-1\) divisor is used for sample covariance because the sample means are estimated from the same data; this is the Bessel correction used to make the sample covariance an unbiased estimator of population covariance.

Covariance is difficult to interpret because its magnitude depends on the units and scales of both variables. **Correlation** standardizes covariance:

\[
r=\frac{s_{xy}}{s_xs_y}
\]

Correlation is unit-free and ranges from -1 to +1, making strength and direction of linear association easier to interpret.

```python
import numpy as np

x = np.array([2, 4, 6, 8], dtype=float)
y = np.array([1, 3, 5, 9], dtype=float)

x_mean = np.mean(x)
y_mean = np.mean(y)
products = (x - x_mean) * (y - y_mean)
manual_sample_covariance = np.sum(products) / (len(x) - 1)

covariance_matrix = np.cov(x, y)  # default ddof=1: sample covariance
correlation_matrix = np.corrcoef(x, y)

print("x mean:", x_mean)
print("y mean:", y_mean)
print("Products:", products)
print("Sum of products:", np.sum(products))
print("Manual sample covariance:", manual_sample_covariance)
print("np.cov(x, y):\n", covariance_matrix)
print("Correlation coefficient:", correlation_matrix[0, 1])
```

## Question 8 — Independence and covariance

The covariance of X and Y is:

\[
\operatorname{Cov}(X,Y)=E(XY)-E(X)E(Y)
\]

If X and Y are independent, then \(E(XY)=E(X)E(Y)\). Therefore:

\[
\operatorname{Cov}(X,Y)=E(X)E(Y)-E(X)E(Y)=0
\]

Thus, **independence implies zero covariance** (when the expectations exist). The reverse is not generally true.

**Example of zero covariance without independence:** Let X take values -1, 0 and 1 with equal probabilities, and define \(Y=X^2\). Y is completely determined by X, so they are dependent. But symmetry gives \(E(X)=0\) and \(E(XY)=E(X^3)=0\), hence covariance is zero.

For the sum of two random variables:

\[
\operatorname{Var}(X+Y)=\operatorname{Var}(X)+\operatorname{Var}(Y)+2\operatorname{Cov}(X,Y)
\]

If X and Y are independent, covariance is zero and the variances add.

```python
import numpy as np

x = np.array([-1, 0, 1], dtype=float)
y = x ** 2

print("X:", x)
print("Y = X^2:", y)
print("Covariance (population):", np.mean((x - x.mean()) * (y - y.mean())))
```

## Question 9 — Bayes' theorem: defective item from two machines

Bayes' theorem:

\[
P(A\mid D)=\frac{P(D\mid A)P(A)}{P(D)}
\]

where:
- **Prior:** Probability before observing the new evidence; here, \(P(A)=0.60\).
- **Likelihood:** Probability of the evidence given a hypothesis; here, \(P(D\mid A)=0.02\).
- **Evidence:** Overall probability of observing the evidence, \(P(D)\).
- **Posterior:** Updated probability after observing the evidence; here, \(P(A\mid D)\).

The supplied data can be arranged as:

| Machine | Prior probability | Defect probability | Joint probability of machine and defect |
|---|---:|---:|---:|
| A | 0.60 | 0.02 | 0.60 × 0.02 = 0.012 |
| B | 0.40 | 0.05 | 0.40 × 0.05 = 0.020 |
| Total | 1.00 | | 0.032 |

Therefore:

\[
P(A\mid D)=\frac{0.60(0.02)}{0.60(0.02)+0.40(0.05)}
=\frac{0.012}{0.032}=\boxed{0.375}
\]

The probability that the defective item was made by machine A is **37.5%**.

```python
p_A = 0.60
p_B = 0.40
p_D_given_A = 0.02
p_D_given_B = 0.05

p_D = p_A * p_D_given_A + p_B * p_D_given_B
p_A_given_D = (p_D_given_A * p_A) / p_D

print("P(defective):", p_D)
print("P(A | defective):", p_A_given_D)
```

## Question 10 — Medical test and the base rate fallacy

Given:
- Disease prevalence: \(P(S)=0.01\)
- Sensitivity: \(P(+\mid S)=0.95\)
- False-positive rate: \(P(+\mid H)=0.05\)
- Healthy probability: \(P(H)=0.99\)

Using Bayes' theorem:

\[
P(S\mid +)=\frac{P(+\mid S)P(S)}
{P(+\mid S)P(S)+P(+\mid H)P(H)}
\]

\[
=\frac{0.95(0.01)}{0.95(0.01)+0.05(0.99)}
=\frac{0.0095}{0.059}
=\boxed{0.1610\text{ or }16.10\%}
\]

The result is much lower than 95% because 95% is the probability of a positive test **given that the person is sick**, not the probability of being sick **given a positive test**. Since the disease is rare, the much larger healthy population can produce many false positives.

**Simulation:** Simulate one million people, assigning disease status using the prevalence and test outcomes using the sensitivity and false-positive rate. The simulated proportion among positive tests should be close to 16.1%, with small random variation.

If prevalence rises to 10%, then:

\[
P(S\mid +)=\frac{0.95(0.10)}{0.95(0.10)+0.05(0.90)}
=\frac{0.095}{0.140}
=\boxed{0.6786\text{ or }67.86\%}
\]

**Base rate fallacy:** This is the mistake of ignoring how common or rare something is in the population when interpreting evidence. A test result must be considered together with the underlying prevalence, not just the test's sensitivity.

```python
import numpy as np

rng = np.random.default_rng(42)
n_people = 1_000_000

# Simulate whether each person has the disease
has_disease = rng.random(n_people) < 0.01

# Simulate test results conditional on disease status
positive_test = np.zeros(n_people, dtype=bool)
positive_test[has_disease] = rng.random(np.sum(has_disease)) < 0.95
healthy = ~has_disease
positive_test[healthy] = rng.random(np.sum(healthy)) < 0.05

positive_count = np.sum(positive_test)
sick_and_positive = np.sum(has_disease & positive_test)
simulated_p_sick_given_positive = sick_and_positive / positive_count

theoretical_1_percent = (0.95 * 0.01) / ((0.95 * 0.01) + (0.05 * 0.99))
theoretical_10_percent = (0.95 * 0.10) / ((0.95 * 0.10) + (0.05 * 0.90))

print("Number of people:", n_people)
print("Positive tests:", positive_count)
print("Sick people with positive tests:", sick_and_positive)
print(f"Simulated P(sick | positive), 1% prevalence: {simulated_p_sick_given_positive:.4%}")
print(f"Theoretical P(sick | positive), 1% prevalence: {theoretical_1_percent:.4%}")
print(f"Theoretical P(sick | positive), 10% prevalence: {theoretical_10_percent:.4%}")
```
